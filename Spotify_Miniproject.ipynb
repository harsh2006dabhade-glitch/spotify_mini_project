```python
import pandas as pd
```

```python
df = df.drop_duplicates(subset=["track_id"], keep="first") #subset means which columns pandas should use , keep first keeps first entry drops other
```

```python
df['track_name'].isna().sum() #isna is is null
```

```python
df=df.dropna(subset=['track_name'])
```

```python
df.count()
```

```python
df['artist_name'].isna().sum()
```

```python
df = df.dropna(subset=["artist_name"])
```

```python
df.count()
```

```python
df["genre"].isna().sum()
```

```python
df['genre'] = df['genre'].fillna('Unknown') #fill na is used to fill the null values
```

```python
df.count()
```

```python
df['popularity'] = pd.to_numeric(df["popularity"], errors="coerce") #coerce is to convert text to numbers without errors
```

```python
df
```

```python
df['popularity'].isna().sum()
```

```python
df['popularity'] = df['popularity'].fillna(df['popularity'].median()) #median is avg value
```

```python
df
```

```python
df['duration_ms'] = pd.to_numeric(df['duration_ms'],errors="coerce")
```

```python
df['duration_ms'].isna().sum()
```

```python
df['duration_ms'] = df['duration_ms'].fillna(df['duration_ms'].median())
```

```python
df
```

```python
df["explicit"] = df["explicit"].replace({
    "True" : "True",
    "False" : "False",
    "true" : "True",
    "false" : "False",
    "TRUE" : "True",
    "FALSE" : "False",
    "Yes":"True",
    "No":"False",
    "yes":"True",
    "no":"False"
})
```

```python
df
```

```python
df['danceability'].isna().sum()
```

```python
df['danceability']=df['danceability'].fillna(df['danceability'].median())
```

```python
df
```

```python
df['danceability'].isna().sum()
```

```python
df.count()
```

```python
df['energy'].isna().sum()
```

```python
df['tempo'].isna().sum()
```

```python
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")
# to_datetime Converttext into actual date format that Panda can understand
# error... if the value cant be converted to date then it is converted to NAT(Not a time),#coerce is to convert text to numbers without errors
```

```python
df
```

```python
df["release_year"]= df["release_date"].dt.year
#dt is date time acesssor
#converts to float instead of int
#and converts nat to nan
```

```python
df
```

```python
df['release_year'].isna().sum()
```

```python
df["release_year"]= df["release_year"].fillna(df["release_year"].mode()[0]).astype(int)
```

```python
df
```

```python
df=df.drop(columns="release_date")
```
